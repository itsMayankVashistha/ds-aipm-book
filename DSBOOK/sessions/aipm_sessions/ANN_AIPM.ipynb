## What is a neuron?

A neuron is a **tiny decision contributor**.

It:
1. Receives inputs
2. Weighs importance
3. Produces an output

```
 x1 ──▶( )
 x2 ──▶( )──▶ output
 x3 ──▶( )
```

## Training the network

### Deep learning and parameter search

How to find the parameters?
* First intuition:  <span class="color-brand">**Gradient descent**</span>
    * But how to apply the algorithm with many coupled and nested functions?
* <span class="color-brand">**Backpropagation**</span> was a breakthrough in research
    * Makes training of very deep networks possible
    * Error is backpropagated through a network (End to Begin)
    * Gradient is calculated step-wise
* Problem: Gradient of the step function is almost always zero
    * What now?

### Non-linear activation functions

<div class="group">  
  <div class="text_70"> 
      
* Make Backpropagation possible
    * <span class="color-brand">Gradient at almost any point in space</span>
* Most common activation functions are
    * <span class="color-brand">Sigmoid</span><span style="display:inline-block; width: 6.7cm;"></span>

    * <span class="color-brand">Tanh</span><span style="display:inline-block; width: 7.85cm;"></span>
    $\hspace{0.5cm} \tanh(z) = \frac{\exp{(z)} - \exp{(-z)}}{\exp{(z)}+ \exp{(-z)}}$
    * <span class="color-brand">ReLU (Rectified Linear Unit)</span> 
    $\hspace{0.5cm} ReLU(z) = \max(0,z)$

## Case study: Schedule risk

**Inputs**:
- Early variance
- Supplier performance
- Team churn

**Output**: Probability of delay

**Value**: Earlier escalation, objective triggers

Notes: Inputs (The Data): The model looks at early variance (early signs of slipping), how suppliers are performing, and how often team members are leaving (team churn).

Output: It provides a specific Probability of Delay.

Product Value: It allows for earlier escalation and creates objective triggers—meaning instead of a person "feeling" a project is late, the data proves it, allowing for faster intervention.

## Case study: Cost overrun

**Inputs**:
- Burn rate trends
- Scope volatility
- Contract type

**Outcome**: Reduced optimism bias

Notes: This application focuses on financial health and accuracy by removing human emotion from budget forecasting.

Inputs (The Data): The model tracks burn rate trends (how fast money is being spent), scope volatility (how much the project requirements change), and the specific contract type.

Outcome: The primary benefit is reduced optimism bias. Humans tend to think things will be cheaper than they are; the model provides a realistic financial outlook based on historical patterns.

## Case study: Benefits realisation

**Inputs**:
- Adoption metrics
- Change saturation
- Capability maturity

**Used for**: Targeted change investment

Notes: This model determines if a product or feature is actually delivering the value it promised to the business.

Inputs (The Data): It analyzes adoption metrics (usage stats), change saturation (whether the organization can handle more new features), and the current capability maturity of the users.

Used For: This data is used for targeted change investment. If the model shows adoption is low because users are overwhelmed, the PM can invest in training or UI improvements rather than just building more features.
