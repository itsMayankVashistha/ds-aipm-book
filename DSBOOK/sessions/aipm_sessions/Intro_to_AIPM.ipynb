### Uncertainty

**A deterministic system** is one in which the occurrence of all events is known with certainty. If the description of the system state at a particular point of time of its operation is given, the next state can be perfectly predicted.


**A probabilistic system** is one in which the occurrence of events cannot be perfectly predicted. Though the behavior of such a system can be described in terms of probability, a certain degree of error is always attached to the prediction of the behavior of the system.

### Heuristics / base model

**A heuristic** is an approach to problem-solving or self-discovery using 'a calculated guess' derived from previous experiences. 

Heuristics are mental shortcuts that ease the cognitive load of making a decision. 

Usually, the opposite process to heuristics is the application of algorithms. Algorithms involve calculated answers and guesswork is eliminated.


Notes: Heuristics are a legitimate way to make decision if they are kept under control. ML allows to go beyond heuristics, but there are overheads involved. In our case : We could assume that grandma cookie monster comes once a week. It’s not 100% right, but it’s not entirely wrong either.

### Machine Learning is a tool to deal with  <u>uncertainty</u> in <u>probabilistic</u> systems

Use it when you have exhausted all other options and **not** because you don't yet have the solution for a deterministic problem.

**As an AI PM:** Be the person who asks "do we actually need ML here?" before the team builds it. Knowing when NOT to use ML is as valuable as knowing when to use it.


<img src="../../images/intro_to_data_science_ml/img_p10_1.png" style="display:block; margin:auto">

### Do <u>NOT</u> solve <u>deterministic</u> problems with Machine Learning 

Using Machine Learning introduces complexity and overheads that can only be justified **if** they are absolutely necessary

**For the PMs:** Before adding AI to a product feature, ask: can this be solved with a simple rule or formula? If yes — don't use ML. Your job is to protect the team from unnecessary complexity.


Notes: Do not solve “Tic Tac Toe” or “Rock paper scissors” with ML

## The AI Product Lifecycle

### The AI Product Lifecycle

<div class="group">
  <div class="text">

Building AI products is an **iterative, cross-functional process** — not a one-time build.

The lifecycle has 5 key phases:

1. **Define** the problem
2. **Assess** the data
3. **Build & Evaluate** the model
4. **Deploy**
5. **Monitor & Iterate**

As an AI PM, you are involved in every phase — not just the beginning and end.

### 1. Define the Problem

<div class="group">
  <div class="text">

**Questions to answer before building anything:**

* What user or business problem are we solving?
* Is this problem well-defined enough to solve with ML?
* What does "good enough" look like? (define success *before* you build)
* What is the cost of a wrong prediction? (false positive vs. false negative)
* Who are the stakeholders and what do they need to know?

**PM's responsibility:**

Write a clear problem statement and get stakeholder sign-off *before* any model development starts.

> "A problem well-stated is a problem half-solved." — Charles Kettering

  </div>
</div>

### 2. Assess the Data

<div class="group">
  <div class="text_70">

**Data is the foundation — and the biggest risk.**

Key questions:

* Do we have data at all? If not, how do we get it?
* Is it **labeled**? Labeling is expensive and slow.
* Is it **representative** of the real world? (watch out for sampling bias)
* Is it **fresh** enough? Stale data = stale model.
* Are there any **legal/privacy** constraints? (GDPR, PII)

**Common failure modes:**

| Issue | Impact |
|---|---|
| Not enough data | Model won't generalise |
| Biased data | Unfair or broken predictions |
| Leaky features | Overly optimistic results |
| No ground truth | Can't evaluate the model |

### 3. Build & Evaluate

<div class="group">
  <div class="text_70">

**Your job is not to build the model — it is to define what "good" means.**

As PM, you:
* Write **acceptance criteria** (what the model must achieve before shipping)
* Define the **baseline**: what does the current solution look like? (rule-based? human? nothing?)
* Distinguish **offline metrics** (measured in the lab) from **online metrics** (measured in production)

| Offline metric | Online metric |
|---|---|
| Accuracy, F1, RMSE | Click-through rate, conversion, churn |
| Measured on held-out test set | Measured on real users in A/B test |
| Fast to compute | Takes time and traffic |

**MVP mindset:** what is the simplest model that moves the needle?

### 4. Deploy

<div class="group">
  <div class="text">

**Shipping a model is not the same as shipping a feature.**

Key decisions for the PM:

* **Build vs. buy** — use an API (OpenAI, Google Vision) or train your own? Cost, latency, control.
* **Rollout strategy** — shadow mode, canary release, or full launch?
* **Rollback plan** — what happens if the model behaves badly in production?
* **User communication** — do users know they are interacting with an AI? (transparency, trust)

**Rollout strategies:**

| Strategy | Description |
|---|---|
| Shadow mode | Model runs silently, output not shown to users yet |
| Canary / A/B | Model shown to a small % of users first |
| Full launch | Rolled out to everyone |

  </div>
</div>

### 5. Monitor & Iterate

<div class="group">
  <div class="text_70">

**Models degrade. The world changes. Your job does not end at launch.**

What to watch:

* **Data drift** — the distribution of incoming data shifts over time
* **Concept drift** — the relationship between input and output changes (e.g. user behaviour shifts)
* **Model degradation** — accuracy, precision, recall drop in production

**When to retrain?**

* On a schedule (e.g. monthly)
* When a performance threshold is breached
* When significant data drift is detected

**The iteration loop:**

Collect feedback → Analyse failures → Improve data or model → Re-evaluate → Re-deploy

> ML products are never "done". Build processes for continuous improvement from day one.
