## How Do Agents Work?

### The Basic Cycle

```
1. OBSERVE (What's happening?)
   ↓
2. THINK (What should I do?)
   ↓
3. ACT (Do something!)
   ↓
4. OBSERVE RESULTS (What happened?)
   ↓
(Repeat continuously)
```

### Real Example: Weather Alert Agent

1. **OBSERVE**: Checks weather forecast every hour
2. **THINK**: "Will it rain tomorrow?"
3. **ACT**: Sends you a notification
4. **OBSERVE**: Continues monitoring

Notes: this is what makes it an agent vs. a simple tool.
Ask: "What other things in your life follow this cycle?"

## ReAct: How Modern Agents Think

**ReAct = Reasoning + Acting**

Modern AI agents combine **thinking** (reasoning) and **doing** (acting) in a loop — just like humans solving problems.

### The ReAct Pattern

1. **Thought** → “I need to find out the weather.”  
2. **Action** → Search for weather in New York.  
3. **Observation** → “It’s 72°F and sunny.”  
4. **Thought** → “Now I can answer the user.”  
5. **Action** → Reply: “It’s 72°F and sunny in New York!”  


 Why It Matters

- Prevents random or unnecessary tool use  
- Makes the agent’s behavior **transparent**  
- Enables **debugging and interpretability**  
- Reflects how **ChatGPT-style agents** and **research copilots** plan and act


 Used In
- ChatGPT Advanced Data Analysis  
- Research & coding assistants  
- Autonomous data and workflow agents

### Building Blocks: LLMs and Tools

### What's an LLM?

**Large Language Model** = The "brain" of modern AI agents

Think of it as:
- A really smart text processor
- Can understand and generate language
- Examples: ChatGPT, Claude, Gemini

**What it does well**:
- Understanding questions
- Planning steps
- Explaining things

**What it CAN'T do alone**:
- Search the internet
- Send emails
- Access databases

They're the reasoning engine, but they need tools to actually DO things.
 LLMs can't natively search the web. They need tools/APIs to do this 
Send emails - LLMs can't directly send emails. They'd need integration with email services through APIs.
LLMs can't directly query databases. They need appropriate APIs, database connectors, or tools to interact with databases.

### What Are Tools?

Tools give agents **capabilities** they don't have naturally.

**Examples of tools**:
```
- Calculator (for math)
- Web search (for finding information)
- Database query (for getting data)
- Email sender (for communication)
- File reader (for documents)
- Weather API (for forecasts)
```

**The magic**: Agent decides WHICH tool to use WHEN

### Putting It Together

```
User Question
    ↓
LLM (Brain) reads question
    ↓
LLM decides
    ↓
Tool runs and gives result
    ↓
LLM reads result
    ↓
LLM generates final answer
```

Notes: Walk through a concrete example:
User: "What's 15% of 876?"
- LLM thinks: "This is math, I need the calculator"
- Calculator tool: 131.4
- LLM responds: "15% of 876 is 131.4"

vs.

User: "What's the weather like?"
- LLM thinks: "I need the weather tool"
- Weather API: "Sunny, 72°F"
- LLM responds: "It's sunny and 72 degrees!"

## Connecting Agents to the World: MCP

### What is MCP?

**Model Context Protocol** = A standardized way for AI agents to access tools and data

Think of it as:
- **Universal adapters for AI agents**
- Like USB ports for your computer - one standard way to connect many things

### The Problem MCP Solves

<div class="group">
  <div class="text_70">
      
**Before MCP:**
Each tool needs custom code → Hard to reuse or share

**With MCP:**
Standardized connections → Plug-and-play tools
  </div>
</div>

Notes: 
 "Remember when every phone had a different charger? Then USB-C became the standard. MCP does this for AI agents."

Ask: "Why would standardization be important when building agents?"

Emphasize: Developers can focus on building the agent's logic, not reinventing connections.

### MCP Servers Are Bridges

MCP servers give LLMs access to:

**Real-time data**

**Actions in external systems**


Notes:
**Real-time data:**
- Databases (customer records, inventory)
- File systems (documents, spreadsheets)  
- Web searches (current information)
- APIs (weather, stock prices)

**Actions in external systems:**
- GitHub (create issues, review code)

- Email (send, read, organize)

### How It Works


## The Role of Memory

Notes: Transition: "Now that we know how agents connect to tools, let's talk about how they remember..."

### Why Memory Matters

**Without memory**:
```
User: "My name is Alex"
Agent: "Nice to meet you!"

User: "What's my name?"
Agent: "I don't know" ❌
```

**With memory**:
```
User: "My name is Alex"
Agent: "Nice to meet you!" [stores: name=Alex]

User: "What's my name?"
Agent: "Your name is Alex!" ✅
```

### Types of Memory

**Short-term (Working) Memory**:
- Remembers current conversation
- Like your brain during a single conversation
- Lost when conversation ends

**Long-term (Persistent) Memory**:
- Remembers across conversations
- Like your brain remembering a friend's birthday
- Stored permanently (until deleted)

**Semantic Memory**:
- General knowledge
- Facts about the world
- Built into the model

Notes: Give practical examples of each:
- Short-term: "As you mentioned earlier in this chat..."
- Long-term: "Last week you told me you prefer..."
- Semantic: "Paris is the capital of France"

### Memory Makes Agents Better

**Example: Shopping Assistant**

Without memory:
- "Show me shoes" every time

With memory:
- Remembers you wear size 42
- Knows you prefer Nike
- Recalls you're looking for running shoes
- Shows relevant options immediately

## Common Challenges

### What Can Go Wrong?

**Problem 1: Agents Get Confused**
- Unclear tool descriptions
- Too many options
- **Solution**: Give clear tool descriptions, be specific


Notes : 
 Simple Example: Calculator Agent

**Step 1: Give Your Agent a Tool**

```python
def calculator(math_problem):
    """A tool that solves math problems"""
    result = eval(math_problem)
    return f"Answer: {result}"
```

**Step 2: Create the Agent**

```python
agent = create_agent(
    tools=[calculator],
    model="gpt-4"
)
```

### Simple Example: Calculator Agent (continued)

**Step 3: Use It!**

```python
agent.ask("What is 25 * 4?")
```

**What Happens Behind the Scenes**:
1. Agent reads your question
2. **Thinks**: "This is math, I have a calculator tool"
3. **Acts**: Uses calculator tool with "25 * 4"
4. **Observes**: Gets result: 100
5. **Responds**: "The answer is 100"

### What Can Go Wrong? 

**Problem 2: Too Many Steps**
- Complex tasks take too long
- Agent loses track
- **Solution**: Keep tasks simple, break into smaller ones

Notes:
"Ask an agent: 'Write a research paper on climate change'
- Too many steps! Agent gets lost.

Instead, break it down:
1. 'Find 3 sources on climate change'
2. 'Summarize these sources'
3. 'Write an outline'
4. 'Write the introduction'

Each step is simple and clear. Agent succeeds!"

### What Can Go Wrong? 

**Problem 3: Agents Make Mistakes**
- Wrong tool selection
- Incorrect reasoning
- **Solution**: Always test thoroughly, have human review

<div class="alert alert-block alert-warning">
Agents are powerful but not perfect - they need supervision!

## Quick Reference

### Agent vs Traditional AI

| Traditional AI | Agent |
|---------------|-------|
| Input → Output (done) | Continuous cycle of sensing and acting |
| Waits for commands | Proactive |
| No memory | Has memory |
| Single-purpose | Multi-step tasks |

### Main Components

- **Sensors**: Gather information
- **Brain**: Think and remember
- **Actuators**: Take actions

### When to Use Agents

**✅ Good use cases:**
- Tasks that need multiple steps
- When you need memory/context
- Automated workflows
- Continuous monitoring

**❌ Not ideal for:**
- Simple one-time questions
- When you need guaranteed accuracy
- Real-time critical decisions (without human oversight)
- Tasks with unclear goals

- Guaranteed accuracy: Agents can make mistakes. For critical calculations or legal/medical advice, don't rely solely on agents.
- Real-time critical decisions: Agents CAN be used for time-sensitive tasks (like trading alerts, emergency notifications) BUT they need human review for critical decisions.
- Tasks with unclear goals: If you can't explain what you want clearly, the agent will struggle.

- OpenAI's official way to build agents. Very reliable and well-maintained, but costs money per API call.
