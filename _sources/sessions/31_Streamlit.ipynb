## Motivation

### The last mile problem

<div class="group">
    <div class="text_70">

* Data Science has a _last mile problem_
    * Analytical work usually in Python, notebooks technical environments
    * Disconnect with people acting on the insights of the analysis
        
    </div>
    <div class="images_30">
    </div>
</div>

The last mile is a concept from logistics — getting the product the final stretch to the customer is often the hardest and most expensive part. In Data Science, the product is insight, and the last mile is getting it to decision-makers.
Obviously, the people who need to act on the insights do not live in the python-notebook world.

* Traditionally, two unsatisfying options:
    * **Static reports** (PDFs, slide decks) — capture a snapshot but lose interactivity
    * **Hand off to web development** — preserves interactivity but takes weeks of translation time

* The fundamental tension:
    * Data Scientists think in **Python**
    * Stakeholders think in **interfaces**

Streamlit resolves this tension by making Python itself the interface. You do not need to learn a second language or a second paradigm.

### What Streamlit is — and is not

* **Is**: The shortest path from a working Python analysis to a shareable, interactive tool
* **Is not**: A general-purpose web framework (no routing, no authentication, no templating)

* It is a niche tool for clear purposes:
    * Interactive dashboards for EDA
    * Proof-of-concept demos for ML models
    * Internal tools for data teams
    * Rapid prototyping before a full web app

It is important to set expectations. Streamlit does one thing extremely well. Knowing its boundaries prevents frustration later.

### Three design principles

Streamlit rests on three principles that shape everything about the framework:

1. **An app is a script, not an architecture**
    * No classes to inherit, no routes to define, no templates to write


2. **Interaction = re-execution**
    * When a user changes an input, the script reruns from the top
    * No callbacks, no event listeners, no wiring

3. **The API should disappear**
    * So small and intuitive that it never becomes the focus
    * Your analysis stays front and center

These three principles are not arbitrary. They are deliberate trade-offs. Each one sacrifices something (flexibility, fine-grained control, feature richness) in exchange for radical simplicity.

## Traditional Web Apps vs. Streamlit: Execution

### Scripts, not applications

<div class="group">
    <div class="text_70">

* In traditional web frameworks (Flask, Django), you define **endpoints and handlers**
    * The framework waits for requests and dispatches them
    * The developer has to think about routing, request/response cycles, state
        
    </div>
    <div class="images_30">
    </div>
</div>

* Streamlit **inverts** this entirely
    * No endpoints, no router
    * Only a script that runs from first line to last
    * The output of each function call appears in the browser in order

This is the single most important concept in Streamlit - the script IS the app ... logic and UI are one. Everything else in the slides build on this insight.

### Reactivity through re-execution

<div class="group">
    <div class="text_70">

* When the user interacts with a widget, Streamlit does something radical:
    * It **reruns the entire script** from the beginning
    * The widget remembers its new value
    * Everything downstream recomputes with the updated value
        
    </div>
    <div class="images_30">
    </div>
</div>

* This is _reactive programming_ achieved through the simplest possible mechanism
    * No dependency graph, no observable pattern, no virtual DOM
    * Mental model: "my script produces a page; when something changes, it produces the page again"

DOM = Document Object Model = the internal roadmap that your browser creates to understand and display a webpage.
When you write HTML, you are writing a script. When the browser reads that script, it builds the DOM; a live, interactive version of your app that exists in the computer's memory.

### A profound consequence

* The developer **never has to think about synchronization**
    * There is no question of "which part of the UI needs to update?"
    * All of it does, because the whole script runs

* This makes reasoning about Streamlit apps remarkably easy
    * Reading an app is like reading a recipe: load, ask, transform, display

* But there is an obvious cost: **performance**
    * If loading data takes 5 seconds, every slider movement takes 5 seconds
    * Streamlit solves this with _caching_ — which we will discuss shortly

The cost of simplicity is the central tension of the framework. Caching is not really optional - especially, when working with bigger models, performance "breaks down" if you don't cache. Very impractical.

## Traditional Web Apps vs. Streamlit: Interface

### Imperative vs. declarative

<div class="group">
    <div class="text_70">

* Most web frameworks are **imperative**
    * Create a DOM element, attach a listener, update on event
* Streamlit is **declarative**
    * Specify *what* should appear, not *how* to construct it

    </div>
    <div class="images_30">
    </div>
</div>

Same distinction as SQL vs. procedural database access
SQL: "give me rows where age > 30" — the database figures out how
Streamlit: "display this DataFrame" — the framework renders an interactive table

### Widgets as expressions

* Traditional UI frameworks: slider is a **component**
    * Instantiate it, place it in a layout, register a callback
* Streamlit: slider is an **expression** (returns a value)

* A single line does three things simultaneously:
    1. **Renders** a slider in the browser
    2. **Returns** the current value to Python
    3. **Triggers** a re-execution when changed

* There is no separation between "the UI part" and "the logic part"
    * They are unified into one _data flow_

This unification of UI and logic is what makes Streamlit apps concise. Not because the framework is magical, but because it eliminated the entire layer of ceremony that web development normally requires. THEY ARE ONE!!

### Type-driven rendering

* `st.write()` inspects the **type** of the object it receives:
    * String → Markdown
    * DataFrame → interactive table
    * Matplotlib figure → chart
    * Dictionary → JSON

* This is _polymorphism_ applied to UI rendering
    * The developer does not need to know the rendering details
    * `st.write()` is the `print()` of interactive applications

## Traditional Web Apps vs. Streamlit: Data Flow

### Top-to-bottom data flow

* Because the script runs top-to-bottom, the data flow has a natural structure:
    * Inputs (widgets) before the logic that depends on them
    * Outputs (charts, tables) after the transformations

* Reading a Streamlit app is like reading a **recipe**
    * Load the data → ask the user for parameters → transform → display
    * No indirection, no callback chains, no hidden state mutations

## Traditional Web Apps vs. Streamlit: Layout

### Spatial organization

* Default layout is **vertical**: each `st.` call adds an element below the previous one
* Several mechanisms to break out of linearity:
    * **Columns** — elements side by side
    * **Sidebar** — persistent control panel on the left
    * **Tabs** — switchable views without scrolling
    * **Expanders** — optional detail behind a collapsible header

* The model is a **document with sections**, not a pixel-perfect canvas
    * Apps look consistent and clean by default
    * Trade-off: no fine-grained CSS control, but no need to think about CSS at all

This trade-off is deliberate. Streamlit's audience is Data Scientists who want their apps to look good enough, not designers who need pixel-perfect layouts. The result is that all Streamlit apps look somewhat similar — which is actually a feature, not a bug.

### The sidebar pattern

<div class="group">
    <div class="text_70">

* The most common layout in Streamlit mirrors professional dashboard tools:
    * **Controls** on the left (sidebar)
    * **Results** in the center (main area)
* This separates **questions** from **answers**
    * Questions: "What parameters should we use?"
    * Answers: "Here is the resulting visualization"
        
    </div>
    <div class="images_30">
    </div>
</div>

## Traditional Web Apps vs. Streamlit: States

### The stateless paradox

* Streamlit's re-execution model is **stateless by design**
    * Every run starts fresh
* But real applications need **memory**:
    * A counter that increments
    * A multi-step form that accumulates answers
    * A chat history that grows

* How can a stateless execution model support stateful behavior?
    * Streamlit provides two mechanisms: _caching_ and _session state_

### Caching: remembering computations

<div class="group">
    <div class="text_70">

* `@st.cache_data` — for computations that produce {data}
    * Loading a CSV, running a query, computing features
    * Stores the return value keyed by function name, source code, and input arguments
    * Conceptually: {memoization} from functional programming
        
    </div>
    <div class="images_30">
    </div>
</div>

* `@st.cache_resource` — for objects that are expensive to {initialize}
    * Database connections, ML models, API clients
    * Stores the actual Python object (shared by reference)

* The mental model:
    * `cache_data` for things you **compute** → each caller gets a copy (safe)
    * `cache_resource` for things you **initialize** → all callers share one instance (efficient)

### Session State: remembering interactions

* `st.session_state` is a dictionary that persists across reruns
    * Per-user and per-browser-tab
    * Solves problems caching cannot:
        * How many times was a button pressed?
        * What did the user select in step 1 of a multi-step form?
        * What is the chat history so far?

* Together, **caching** and **session state** create the illusion of continuity
    * Even though the underlying mechanism is repeated full re-execution
    * The user never notices the script is rerunning from scratch

Session state is the escape hatch for when the stateless rerun model is not enough. Caching remembers computations. Session state remembers interactions. Together they make Streamlit apps feel like persistent applications.

## From Notebook to App

### The conceptual distance

<div class="group">
    <div class="text_70">

* A typical notebook workflow:
    * Load data → clean → explore → model → evaluate
    * Each step is a cell, the notebook is sequential
* A Streamlit app has the **same structure**
    * Replace hardcoded parameters with widgets
    * Replace `display()` with `st.write()`
    * Add caching for expensive steps
        
    </div>
    <div class="images_30">
    </div>
</div>

* No restructuring, no new architecture
    * Just a change in how **inputs and outputs** are handled
    * The analysis stays the same

Remember: WE ARE NOT WEB DEVELOPERS. And that's fine.

### When the distance grows

* Streamlit's simplicity has limits:
    * User authentication
    * Persistent data storage
    * Complex navigation with many pages
    * Fine-grained visual design

* For these cases, Streamlit is best understood as a **prototyping tool**:
    * Build the proof of concept in Streamlit
    * Validate the idea with stakeholders
    * Invest in a full web app (Flask, Django) if the idea proves its value

Notes: Knowing the limits of a tool is as important as knowing its strengths. Streamlit excels at rapid iteration and communication. It is not designed for production-grade software engineering.

## Deployment

### Closing the communication gap

* An app on `localhost` is not much better than a notebook
    * {Deployment} is what actually closes the communication gap

* **Streamlit Community Cloud**: the simplest path
    * Connect a GitHub repo with a `requirements.txt`
    * App is live at a public URL within minutes
    * Free for public repositories

* **Any Python host** works for production:
    * Docker containers, cloud VMs, platform-as-a-service, Kubernetes
    * No build step, no compiled frontend, no separate API server
    * Just a Python process serving a web page

## Conclusion

### Key Concepts

* Streamlit addresses the **last mile problem**: getting insights from notebooks to stakeholders

* The **execution model** is a script that reruns top-to-bottom on every interaction: radical simplicity

* The interface is **declarative**: specify *what* to display, not *how* to render it

* Widgets are **expressions** that return values: unifying UI and logic

* **Caching** compensates for re-execution cost; **session state** provides memory across reruns

* The gap between notebook and app is a change in **inputs and outputs**, not a change in paradigm
