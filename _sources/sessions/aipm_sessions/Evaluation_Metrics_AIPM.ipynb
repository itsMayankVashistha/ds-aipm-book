### Which Metrics Can We Use to Evaluate Model Performance?
        
R2 (R-squared)


$${R}^{2}=1-\frac{R S S}{T S S}=1-\frac{\sum_{i=1}^{n}(y_{i}-{\hat{y}}_{i})^{2}}{\sum_{i=1}^{n}(y_{i}-{\bar{y}})^{2}}=1-\frac{\sum_{i=1}^{n}e_{i}^{2}}{\sum_{i=1}^{n}(y_{i}-{\bar{y}})^{2}}$$    

MSE (**M**ean **S**quare **E**rror)


$$MSE={\frac{1}{n}}\sum_{i=1}^{n}\left(y_{i}-{\hat{y}}_{i}\right)^{2}$$

RMSE (**R**oot **M**ean **S**quare **E**rror)


$$RMSE = \sqrt{MSE}$$

MAPE (**M**ean **A**bsolute **P**ercentage **E**rror)


$${MAPE}=\frac{1}{n}\sum_{i=1}^{n}\frac{|y_{i}-\hat{y}_{i}|}{\mathrm{max}(\epsilon,\left|y_{i}\right|)}$$

Notes: MAPE is scale-invariant and easier to communicate as a percentage error. Mention that $\epsilon$ is a small positive value used to avoid division by zero when true values are zero, but also note that near-zero true values can still make MAPE unstable and overly sensitive.

## Going Beyond Aggregated Metrics
All the performance metrics we have seen today are aggregated metrics.

They help determine whether a model has learned well from a dataset or needs
improvement.

**Next step**: \
Examine results and errors to understand why and how the model is failing or succeeding.

<span class="color-brand">Why: validation and iteration</span>

<div class="alert alert-block alert-info">
Performance metrics can be deceptive: on highly imbalanced datasets, a classifier can reach very high accuracy without real predictive power.
</div>



### Validate Your Model → Inspect How It Performs
There are many ways to do this. The goal is to compare your model's predictions with the actual data (targets and/or features).

**Regression**: \
Examine residuals, perform exploratory data analysis (EDA), and look for patterns or outliers.

**Classification**: \
Start with a confusion matrix to analyze how well predicted classes match true classes.


### Product Goal → Metric Choice (Quick Guide)

Use one **primary metric** for decisions and one or two **guardrails** to control risk.

**PR AUC** = area under the precision-recall curve. It summarizes precision-recall trade-offs across thresholds and is especially informative when the positive class is rare.

**FNR** = false negative rate, the share of actual positives the model misses.

| Product goal | Primary metric | Guardrail metric(s) | Why this works |
|---|---|---|---|
| Catch as many true cases as possible (e.g., fraud/cancer pre-screening) | Recall (TPR) | Precision, FPR | Maximizes detection while controlling false alarms |
| Avoid false alarms (e.g., spam filter for important inboxes) | Precision | Recall, FNR | Keeps positive predictions trustworthy |
| Balance misses and false alarms when both matter | F1-score | Precision, Recall | Single score balancing both error types |
| Compare classifier ranking quality across thresholds | ROC AUC and, when positives are rare, PR AUC as a complementary view | Chosen operating threshold, class balance context | Good for model comparison before threshold selection |
| Communicate baseline classification performance quickly | Accuracy | Class balance check, Recall | Simple, but only meaningful when classes are not highly imbalanced |
| Regression forecast quality in original units | RMSE | MAPE, error-by-segment | RMSE is interpretable in target units and penalizes large errors. Use MAPE cautiously when true values can be near zero, because percentage errors become unstable. |

<div class="alert alert-block alert-info">
<b>Decision pattern:</b> Objective → Primary metric → Guardrails → Threshold policy.
</div>

### Choosing Metrics for Product Decisions (Wrap-up)

* **Business objective**: Define the outcome to optimize (e.g., catch fraud, reduce churn).
* **Primary metric**: Pick one decision metric (e.g., Recall at threshold, PR AUC, RMSE).
* **Guardrail metric(s)**: Add metrics that prevent harmful trade-offs (e.g., precision, FPR, review cost).
* **Threshold policy**: Set and revisit thresholds based on estimated benefit vs. cost.

<div class="alert alert-block alert-info">
Use this sequence in stakeholder updates: objective → primary metric → guardrails → threshold decision.
</div>
